```python
df_analysis = df[df['booking_status'].isin(['Confirmed', 'Cancelled'])].copy()
```

```python
# Create a binary flag: 1 for Cancelled, 0 for Confirmed
df_analysis['is_cancelled'] = df_analysis['booking_status'].apply(lambda x: 1 if x == 'Cancelled' else 0)
```

```python
# Calculate Overall Cancellation Rate
overall_rate = df_analysis['is_cancelled'].mean()
print(f"Overall Cancellation Rate: {overall_rate:.2%}\n")
```

```python
# 3. Calculate Percentages by Category
def get_cancellation_stats(dataframe, column):
    stats = dataframe.groupby(column)['is_cancelled'].mean().sort_values(ascending=False)
    return stats
```

```python
# Get rates for key features
channel_rates = get_cancellation_stats(df_analysis, 'booking_channel')
room_rates = get_cancellation_stats(df_analysis, 'room_type')
star_rates = get_cancellation_stats(df_analysis, 'star_rating')

print("--- Cancellation Rate by Booking Channel ---")
print(channel_rates.map('{:.2%}'.format))
print("\n--- Cancellation Rate by Room Type ---")
print(room_rates.map('{:.2%}'.format))
print("\n--- Cancellation Rate by Star Rating ---")
print(star_rates.map('{:.2%}'.format))

# 4. Visualization
# Setting the visual style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(15, 10))
```

## 3 - Are there seasonal or temporal trends influencing booking values or stay lengths?


```python


# Load and prepare dates
df = pd.read_csv('Hotel_bookings_final.csv')
df['check_in_date'] = pd.to_datetime(df['check_in_date'], errors='coerce')
df['check_out_date'] = pd.to_datetime(df['check_out_date'], errors='coerce')

# Calculate Stay Length and filter for Confirmed bookings
df['stay_length'] = (df['check_out_date'] - df['check_in_date']).dt.days
confirmed = df[df['booking_status'] == 'Confirmed'].copy()

# Group by Month-Year
confirmed['month_year'] = confirmed['check_in_date'].dt.to_period('M').astype(str)
trends = confirmed.groupby('month_year').agg({
    'booking_value': 'mean',
    'stay_length': 'mean'
}).reset_index()

# Plotting
plt.figure(figsize=(12, 6))
sns.lineplot(data=trends, x='month_year', y='booking_value', marker='o', label='Avg Value')
plt.xticks(rotation=45)
plt.title('Monthly Booking Value Trends')
plt.savefig('booking_value_trends.png')
```

Observation: Because stay length is flat while booking values are rising, it indicates that the Average Daily Rate (ADR) is increasing over time. The hotels are successfully charging more per night rather than simply hosting guests for longer durations.

## 3. Business Recommendations

## 1 - Suggest strategies to reduce cancellations.
## 2 - Recommend ways to improve profitability and increase repeat bookings.
## 3 - Identify opportunities to optimize pricing, promotions, or channel strategy.



1. Strategies to Reduce Cancellations
The data shows a high concentration of cancellations in specific segments.

Stricter Policies for Travel Agents: With a 29.7% cancellation rate (the highest among all channels), bookings from Travel Agents are highly volatile. Implementing a non-refundable deposit or a stricter 48-hour cancellation window for this channel specifically would mitigate risk.

Lock in Standard Room Bookings: Standard rooms have a 25.2% cancellation rate, likely due to price-sensitive "deal hunters" switching to cheaper alternatives. Introduce a "Pre-paid/Non-refundable" price tier for Standard rooms at a 5-10% discount to secure the revenue upfront.

Address High-Star Rating Volatility: 5-star properties have higher cancellation rates than 2-star ones. This suggests luxury travelers are "comparison shopping" even after booking. Implement personalized follow-up emails or "Exclusive Perk" confirmations (e.g., complimentary breakfast or spa credit) for 5-star bookings to increase the "switching cost" for the guest.

2. Improving Profitability and Repeat Bookings
The dataset reveals a highly loyal customer base, with some IDs booking over 90 times.

Maximize the Web Channel: The Web channel is the "Gold Mine." It has the highest average profit ($6,985) and the lowest cancellation rate (17.6%). Redirect marketing spend from Travel Agent commissions toward SEO and direct-booking incentives to shift traffic to this channel.

Tiered Loyalty Program: Since the top customers are booking frequently, implement a tiered loyalty system (Silver, Gold, Platinum). Offer guaranteed room upgrades for Platinum members in the Suite/Deluxe categories, which have significantly lower cancellation rates (~16-18%).

Upsell at the Right Moment: Since "Stay Length" is consistently flat at 4 days while prices are rising, profitability can be improved by upselling "add-ons" (early check-in, late check-out, or meal packages) rather than trying to extend the stay duration.

3. Optimizing Pricing, Promotions, and Channel Strategy
Dynamic Seasonal Pricing: Average booking values peak significantly in March, April, and May (reaching over $27,000). The hotel should implement Yield Management—gradually increasing rates as the lead time decreases during these peak spring months to capture the "last-minute" premium.

Pivot Coupon Strategy: Coupon users have a higher cancellation rate (22.6%). Instead of general discount codes, transition to "Value-Add" coupons (e.g., $50 food & beverage credit). These are harder to compare against other hotels' raw price discounts and encourage on-property spending.

Mobile App Optimization: The Mobile App is a high-volume channel with a moderate cancellation rate (21.5%). Use Push Notifications to offer "Mobile-Only" check-in or digital keys to "lock in" the guest’s commitment as the check-in date approaches.
